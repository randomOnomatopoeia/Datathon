```python
import pandas as pd
import altair as alt

alt.data_transformers.enable('vegafusion')
```

```python
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')
merged = pd.merge(sales,items,on = "item_name")
merged
```

```python
merged["date"] = pd.to_datetime(merged["date"])

merged["year_month"] = merged["date"].dt.strftime('%B %Y')

orderTypes_monthly = merged[["transaction_type", "year_month"]].reset_index()

orderTypes_monthly = orderTypes_monthly.groupby(['transaction_type', 'year_month']).size().reset_index(name='number_orders')

orderTypes_monthly['year_month'] = pd.to_datetime(orderTypes_monthly['year_month'], format='%B %Y')

orderTypes_monthly = orderTypes_monthly.sort_values(by='year_month')

orderTypes_monthly_graph = alt.Chart(orderTypes_monthly).mark_line().encode(
    x = alt.X("year_month").title("Month"),
    y = alt.Y("number_orders").title("Number of Orders"),
    color = alt.Color("transaction_type").title("Type"),
).properties(
    title = alt.Title("Number of Takeout Orders vs Dine-In Orders Over Time")
)

orderTypes_monthly_graph
```
