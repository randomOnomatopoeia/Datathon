```python
import pandas as pd
import altair as alt

alt.data_transformers.enable('vegafusion')
```

```python
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')
merged = pd.merge(sales,items,on = "item_name")
merged
```

```python
import pandas as pd
import altair as alt

merged["date"] = pd.to_datetime(merged["date"])

merged["year_month"] = merged["date"].dt.to_period("M").dt.to_timestamp()

orderTypes_monthly = (
    merged.groupby(["transaction_type", "year_month"])
          .size()
          .reset_index(name="number_orders")
)

numberCustomersTotal = (
    merged.groupby("year_month")
          .size()
          .reset_index(name="order_count")
)

orderTypes_monthly_graph = alt.Chart(orderTypes_monthly).mark_line().encode(
    x = alt.X("year_month:T", title="Month"),
    y = alt.Y("number_orders", title="Number of Orders"),
    color="transaction_type"
)

numberCustomersTotal_graph = alt.Chart(numberCustomersTotal).mark_line().encode(
    x = alt.X("year_month:T", title="Month"),
    y = alt.Y("order_count", title="Total Orders")
)

final_graph = orderTypes_monthly_graph + numberCustomersTotal_graph
final_graph
```
