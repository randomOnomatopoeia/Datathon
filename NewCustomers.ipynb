```python
import pandas as pd
import altair as alt

alt.data_transformers.enable('vegafusion')
```

```python
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')
merged = pd.merge(sales,items,on = "item_name")
merged
```

```python
newCustomers = merged.drop_duplicates("customer_id")

newCustomers["date"] = pd.to_datetime(newCustomers["date"])

newCustomers["year_month"] = newCustomers["date"].dt.strftime('%B %Y')

newCustomers_monthly = newCustomers[["customer_id", "year_month"]].reset_index()

newCustomers_monthly = newCustomers_monthly.groupby(['year_month']).size().reset_index(name='number_new_customers')

newCustomers_monthly['year_month'] = pd.to_datetime(newCustomers_monthly['year_month'], format='%B %Y')

newCustomers_monthly = newCustomers_monthly.sort_values(by='year_month')

newCustomers_monthly_graph = alt.Chart(newCustomers_monthly).mark_line().encode(
    x = alt.X("year_month").title("Month"),
    y = alt.Y("number_new_customers").title("Number of New Customers"),
).properties(
    title = alt.Title("Number of New Customers Over Time")
)

newCustomers_monthly_graph
```
