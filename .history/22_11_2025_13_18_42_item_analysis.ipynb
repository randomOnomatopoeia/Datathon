```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import altair as alt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn import set_config
alt.data_transformers.enable('vegafusion')
```

```python

# Load a CSV file named 'data.csv' into a DataFrame
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')

print(items.head())
print(sales.head())

total_data = pd.merge(sales,items,on = "item_name")
total_data = total_data.assign(date_time = pd.to_datetime(total_data["date"] + " " + total_data["time"]))
total_data = total_data.assign(profit = total_data["price"] - total_data["production_cost"])
total_data
```

```python

#profit per item
item_profit = total_data.groupby("item_name")["profit"].sum().reset_index()

prof_by_item = alt.Chart(item_profit).mark_bar(point = True).encode(
    x = alt.X("item_name").title("time"),
    y = alt.Y("profit").title("profit (dollars)")
)
prof_by_item

items = item_profit["item_name"].tolist()
item_time = total_data.groupby(["item_name","date_time"])["profit"].sum().reset_index()

monthly_items = total_data.groupby([pd.PeriodIndex(total_data['date_time'], freq="M"), "item_name"])["profit"].sum().reset_index()
#monthly_items["times"] = monthly_items.index.to_timestamp()

monthly_items
item_time
```

```python
monthly_items = total_data.groupby([pd.PeriodIndex(total_data['date_time'], freq="M"), "item_name"])["profit"].sum().reset_index()
monthly_items["date_time"] = monthly_items["date_time"].dt.to_timestamp("M")
item_month_profit = alt.Chart(monthly_items).mark_area(point=True).encode(
    x = alt.X("date_time:T").title("Time"),
    y = alt.Y("profit").title("profit"),
    color = alt.Color("item_name").title("Item")
)
item_month_profit
```

```python
item_month_facet = alt.Chart(monthly_items).mark_area(point=True).encode(
    x = alt.X("date_time:T").title("Time"),
    y = alt.Y("profit").title("profit"),
    color = alt.Color("item_name").title("Item")
).facet("item_name", columns = 3)
item_month_facet
```

```python
#did increased merch advertising help? 
merch_data = monthly_items[(monthly_items["item_name"] == "Tote Bag") | (monthly_items["item_name"] == "Reusable Coffee Cup")]
merch_data 
merch_month_facet = alt.Chart(merch_data).mark_area(point=True).encode(
    x = alt.X("date_time:T").title("Time"),
    y = alt.Y("profit").title("profit"),
    color = alt.Color("item_name").title("Item")
)

rules = alt.Chart(pd.DataFrame({
  'Date': ['2023-11-1'],
  'color': ['red']
})).mark_rule().encode(
  x='Date:T',
  color=alt.Color('color:N', scale=None)
)

merch_month_facet = (merch_month_facet + rules).facet("item_name")
merch_month_facet 
```
