```python
import pandas as pd
import altair as alt

alt.data_transformers.enable('vegafusion')
```

```python
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')
merged = pd.merge(sales,items,on = "item_name")

print(sales["time"].max())
print(sales["time"].min())
```

```python
sales["time"] = pd.to_datetime(sales["time"])

time_split = pd.to_datetime("11:30:00").time()

morningPerson = sales[sales["time"].dt.time < time_split].reset_index()

morningPerson
```

```python
afternoonPerson = sales[sales["time"].dt.time >= time_split].reset_index()

afternoonPerson
```

```python
morningPerson["date"] = pd.to_datetime(morningPerson["date"])

morningPerson["year_month"] = morningPerson["date"].dt.strftime('%B %Y')

morning_monthly = morningPerson.drop(columns=["index", "date", "time", "transaction_type", "own_cup", "surcharge", "customer_id"])

morning_monthly = morning_monthly.groupby(['item_name', 'year_month']).size().reset_index(name='purchases')

morning_monthly['year_month'] = pd.to_datetime(morning_monthly['year_month'], format='%B %Y')

morning_monthly = morning_monthly.sort_values(by='year_month')

morning_monthly_graph = alt.Chart(morning_monthly).mark_line().encode(
    x = alt.X("year_month").title("Month"),
    y = alt.Y("purchases").title("Number of Purchases"),
    color = alt.Color("item_name").title("Item")
).facet(
    facet = alt.Facet("item_name").title("Number of Items Purchased over Time for Morning People"),
    columns = 3
)

morning_monthly_graph
```

```python
afternoonPerson["date"] = pd.to_datetime(afternoonPerson["date"])

afternoonPerson["year_month"] = afternoonPerson["date"].dt.strftime('%B %Y')

afternoon_monthly = afternoonPerson.drop(columns=["index", "date", "time", "transaction_type", "own_cup", "surcharge", "customer_id"])

afternoon_monthly = afternoon_monthly.groupby(['item_name', 'year_month']).size().reset_index(name='purchases')

afternoon_monthly['year_month'] = pd.to_datetime(afternoon_monthly['year_month'], format='%B %Y')

afternoon_monthly = afternoon_monthly.sort_values(by='year_month')

afternoon_monthly_graph = alt.Chart(afternoon_monthly).mark_area().encode(
    x = alt.X("year_month").title("Month"),
    y = alt.Y("purchases").title("Number of Purchases"),
    color = alt.Color("item_name").title("Item")
)

afternoon_monthly_graph
```
