```python
import pandas as pd
```

```python
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')
merged = pd.merge(sales,items,on = "item_name")

print(sales["time"].max())
print(sales["time"].min())
```

```python

sales["time"] = pd.to_datetime(sales["time"])

time_split = pd.to_datetime("11:30:00").time()

morningPerson = sales[sales["time"].dt.time < time_split].reset_index()

morningPerson
```

```python
afternoonPerson = sales[sales["time"].dt.time >= time_split].reset_index()

afternoonPerson
```

```python
morning_monthly = morningPerson.drop(columns="date").groupby(morningPerson["time"].dt.month)

morning_monthly
```
