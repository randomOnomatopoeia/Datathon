```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import altair as alt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn import set_config
alt.data_transformers.enable('vegafusion')
```

```python

# Load a CSV file named 'data.csv' into a DataFrame
items = pd.read_csv('items.csv')
sales = pd.read_csv('sales.csv')

print(items.head())
print(sales.head())

total_data = pd.merge(sales,items,on = "item_name")
total_data = total_data.assign(date_time = pd.to_datetime(total_data["date"] + " " + total_data["time"]))
total_data = total_data.assign(profit = total_data["price"] - total_data["production_cost"])
total_data = total_data.assign(profit_percent = total_data["production_cost"] / total_data["profit"])
total_data
```

```python

#profit per item
item_profit = total_data.groupby("item_name")["profit"].sum().reset_index()

prof_by_item = alt.Chart(item_profit).mark_bar(point = True).encode(
    x = alt.X("item_name").title("time"),
    y = alt.Y("profit").title("profit (dollars)")
)
prof_by_item

items = item_profit["item_name"].tolist()
item_time = total_data.groupby(["item_name","date_time"])["profit"].sum().reset_index()

monthly_items = total_data.groupby([pd.PeriodIndex(total_data['date_time'], freq="M"), "item_name"])["profit"].sum().reset_index()
#monthly_items["times"] = monthly_items.index.to_timestamp()

monthly_items
item_time
```

```python
# find recurring customers
recur = total_data.copy()
recur['date_time'] = pd.to_datetime(recur['date_time'])
recur = recur.sort_values(['customer_id', 'date_time']).reset_index(drop=True)

# Identify recurring customers (at least 2 transactions within 50 days)
recurring_customers_set = set()

for cust_id, group in recur.groupby('customer_id'):
    dates = group['date_time'].dt.normalize().drop_duplicates().values
    start = 0
    for end in range(1, len(dates)):
        diff = (dates[end] - dates[start]).astype('timedelta64[D]').astype(int)
        while diff > 30:
            start += 1
            diff = (dates[end] - dates[start]).astype('timedelta64[D]').astype(int)
        if end - start >= 1:
            recurring_customers_set.add(cust_id)
            break

recurring_customers_list = list(recurring_customers_set)
print(f"Found {len(recurring_customers_list)} recurring customers")

# Filter transactions for recurring customers
recur_recurring = recur[recur['customer_id'].isin(recurring_customers_list)].copy()
recur_recurring['month'] = recur_recurring['date_time'].dt.to_period('M').dt.to_timestamp()

# Aggregate monthly
recurring_per_month = recur_recurring.groupby('month')['customer_id'].nunique().reset_index()
recurring_per_month.rename(columns={'customer_id': 'num_recurring_customers'}, inplace=True)

# Altair visualization
chart_recurring = alt.Chart(recurring_per_month).mark_line(point=True).encode(
    x='month:T',
    y='num_recurring_customers:Q',
    tooltip=['month:T', 'num_recurring_customers:Q']
).properties(
    title='Number of Recurring Customers Over Time',
    width=800,
    height=400
)

chart_recurring
```

```python
# Identify non-recurring customers
all_customers = set(recur['customer_id'].unique())
non_recurring_customers_set = all_customers - set(recurring_customers_list)

# Filter transactions
df_non_recurring = recur[recur['customer_id'].isin(non_recurring_customers_set)].copy()
df_non_recurring['month'] = df_non_recurring['date_time'].dt.to_period('M').dt.to_timestamp()

# Aggregate monthly
non_recurring_per_month = df_non_recurring.groupby('month')['customer_id'].nunique().reset_index()
non_recurring_per_month.rename(columns={'customer_id': 'num_non_recurring_customers'}, inplace=True)

# Altair visualization
chart_non_recurring = alt.Chart(non_recurring_per_month).mark_line(point=True).encode(
    x='month:T',
    y='num_non_recurring_customers:Q',
    tooltip=['month:T', 'num_non_recurring_customers:Q']
).properties(
    title='Number of Non-Recurring Customers Over Time',
    width=800,
    height=400
)
```
